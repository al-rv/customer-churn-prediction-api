```python
import joblib

loaded_model = joblib.load(
    "../models/churn_model.joblib"
)
```

```python
import pandas as pd

customer = pd.DataFrame([{
    "gender": "Female",
    "SeniorCitizen": 0,
    "Partner": "No",
    "Dependents": "No",
    "tenure": 5,
    "PhoneService": "Yes",
    "MultipleLines": "No",
    "InternetService": "Fiber optic",
    "OnlineSecurity": "No",
    "OnlineBackup": "No",
    "DeviceProtection": "No",
    "TechSupport": "No",
    "StreamingTV": "Yes",
    "StreamingMovies": "Yes",
    "Contract": "Month-to-month",
    "PaperlessBilling": "Yes",
    "PaymentMethod": "Electronic check",
    "MonthlyCharges": 95.0,
    "TotalCharges": 475.0
}])
```

```python
churn_probability = loaded_model.predict_proba(customer)[0, 1]

churn_probability
```

```python
threshold = 0.30

prediction = int(churn_probability >= threshold)

prediction
```
