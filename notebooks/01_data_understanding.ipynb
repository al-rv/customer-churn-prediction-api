```python
df["TotalCharges"].dtype
```

Conceptually now NaN → 0 this isn't:
"Whenever data is missing, replace it with zero."

It's:
"For these specific observations, our investigation suggests zero has a reasonable business interpretation."

That distinction is important.


```python
df["TotalCharges"] = df["TotalCharges"].fillna(0)
```
